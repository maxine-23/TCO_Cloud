# TCO Material Recommendation – Exploratory Analysis

This notebook explores the **demonstration / synthetic** dataset used by the Random Forest pipeline.
All numeric values are relative scores created for pipeline illustration only; they are **not** experimental measurements.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA_PATH = Path("../data/tco_materials.csv")
df = pd.read_csv(DATA_PATH)
df
```

## Basic statistics

```
df.describe().round(2)
```

## Correlation heatmap of the eight properties + suitability

```
plt.figure(figsize=(10, 8))
sns.heatmap(df.drop(columns=["Material"]).corr(), annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.title("Correlation Matrix (Demonstration Data)")
plt.tight_layout()
plt.show()
```

## Suitability ranking (synthetic scores)

```
df_sorted = df.sort_values("Suitability_Score", ascending=False)
plt.figure(figsize=(8, 5))
plt.bar(df_sorted["Material"], df_sorted["Suitability_Score"], color="steelblue")
plt.ylabel("Synthetic Suitability Score")
plt.title("Demonstration Suitability Scores by Material")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
df_sorted[["Material", "Suitability_Score"]]
```
